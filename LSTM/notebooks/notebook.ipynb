```python
from __future__ import annotations


import json
import os
import random
from datetime import datetime, timedelta
from pathlib import Path
from typing import Any, Dict, List, Optional, Tuple, Union


import joblib
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.metrics import (
    mean_absolute_error,
    mean_absolute_percentage_error,
    mean_squared_error,
)
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras import Sequential
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.layers import Dense, Dropout, LSTM




def set_seeds(seed: int = 42) -> None:
    """
    Define seeds globais para reprodutibilidade de experimentos.

    Contexto:
        Garante resultados determinísticos para operações pseudoaleatórias
        envolvendo Python, NumPy e TensorFlow.

    Args:
        seed: Valor inteiro utilizado como semente global.

    Returns:
        None.
    """
    os.environ["PYTHONHASHSEED"] = str(seed)
    random.seed(seed)
    np.random.seed(seed)


    try:
        tf.keras.utils.set_random_seed(seed)
    except Exception:
        tf.random.set_seed(seed)




set_seeds(42)
print("TensorFlow:", tf.__version__)

```

```python
DATA_SOURCE = "csv"

def _find_project_root(start: Optional[Path] = None) -> Path:
    """
    Localiza o diretório raiz do projeto a partir de um ponto inicial.

    Contexto:
        Procura por um arquivo "pyproject.toml" subindo na hierarquia de
        diretórios, permitindo resolver caminhos relativos de forma estável.

    Args:
        start: Caminho inicial para busca. Se omitido, usa o diretório atual.

    Returns:
        Caminho absoluto do diretório raiz encontrado ou do diretório atual
        caso nenhum "pyproject.toml" seja localizado.
    """
    cur = (start or Path.cwd()).resolve()
    for p in (cur, *cur.parents):
        if (p / "pyproject.toml").exists():
            return p
    return cur


PROJECT_ROOT = _find_project_root()


def _resolve_path(p: str) -> str:
    """
    Resolve um caminho possivelmente relativo em relação ao `PROJECT_ROOT`.

.
    Contexto:
        Garante que caminhos de dados, artefatos e saídas sejam sempre
        absolutos, independente de onde o notebook seja executado.

    Args:
        p: Caminho de arquivo ou diretório, absoluto ou relativo.

    Returns:
        Caminho absoluto correspondente ao argumento informado.
    """
    pp = Path(p).expanduser()
    if not pp.is_absolute():
        pp = (PROJECT_ROOT / pp).resolve()
    return str(pp)


DATA_PATH = _resolve_path("data/finance_data.csv")
DATE_COL = os.getenv("TRAIN_DATE_COL", "Date").strip()
TARGET_COL = os.getenv("TRAIN_TARGET_COL", "PETR4.SA").strip()


SYMBOL = os.getenv("TRAIN_SYMBOL", "PETR4.SA").strip()
if DATA_SOURCE == "yfinance":
    DATE_COL = "Date"
    TARGET_COL = SYMBOL


START_DATE = os.getenv(
    "TRAIN_START_DATE",
    (datetime.today() - timedelta(days=300)).strftime("%Y-%m-%d"),
).strip()
END_DATE = os.getenv("TRAIN_END_DATE", datetime.today().strftime("%Y-%m-%d")).strip()




def _env_int(name: str, default: int) -> int:
    """
    Lê uma variável de ambiente e converte para inteiro com valor padrão.

    Contexto:
        Permite parametrizar o treinamento via variáveis de ambiente,
        tratando conversões e valores ausentes de forma segura.

    Args:
        name: Nome da variável de ambiente a ser lida.
        default: Valor inteiro utilizado caso a variável não exista ou
            não possa ser convertida para número inteiro.

    Returns:
        Valor inteiro obtido da variável de ambiente ou o `default`.
    """
    v = os.getenv(name)
    if v is None or str(v).strip() == "":
        return default
    try:
        return int(float(str(v).strip()))
    except Exception:
        return default




TRAIN_MAX_ROWS = _env_int("TRAIN_MAX_ROWS", 0)


TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15


TRAIN_GRID_MODE = os.getenv("TRAIN_GRID_MODE", "full").strip().lower()


if TRAIN_GRID_MODE in ("fast", "quick", "small"):
    HYPERPARAM_GRID = {
        "lstm_units": [32],
        "dropout_rate": [0.2],
        "batch_size": [32],
        "epochs": [10],
        "window_size": [30],
        "learning_rate": [1e-3],
    }
elif TRAIN_GRID_MODE in ("medium", "mid"):
    HYPERPARAM_GRID = {
        "lstm_units": [32, 64],
        "dropout_rate": [0.2],
        "batch_size": [32],
        "epochs": [20],
        "window_size": [30, 60],
        "learning_rate": [1e-3],
    }
else:
    HYPERPARAM_GRID = {
        "lstm_units": [32, 64, 96],
        "dropout_rate": [0.2, 0.3],
        "batch_size": [16, 32],
        "epochs": [30, 50],
        "window_size": [30, 60],
        "learning_rate": [1e-3],
    }


HYPERPARAM_GRID["window_size"] = [30]


EARLY_STOPPING = dict(
    monitor="val_loss",
    patience=_env_int("TRAIN_EARLY_STOPPING_PATIENCE", 10),
    restore_best_weights=True,
)


ARTIFACT_PATH = _resolve_path(os.getenv("TRAIN_ARTIFACT_PATH", "./src/artifacts/best_lstm_artifact.pkl").strip())


OUTPUT_PATH = _resolve_path(os.getenv("TRAIN_OUTPUT_PATH", "./prediction_output.json").strip())


print(
    "Config:",
    {
        "DATA_SOURCE": DATA_SOURCE,
        "TARGET_COL": TARGET_COL,
        "TRAIN_MAX_ROWS": TRAIN_MAX_ROWS,
        "TRAIN_GRID_MODE": TRAIN_GRID_MODE,
        "DATA_PATH": DATA_PATH,
        "WINDOW_SIZE": HYPERPARAM_GRID["window_size"],
    },
)
```

### Load data

```python
def load_data() -> pd.DataFrame:
    """
    Carrega os dados da série temporal a partir de CSV local ou Yahoo Finance.

    Contexto:
        Abstrai a origem dos dados permitindo treinar o modelo tanto com
        arquivos locais quanto com download dinâmico via `yfinance`.

    Returns:
        DataFrame contendo a série histórica completa, possivelmente limitada
        por `TRAIN_MAX_ROWS` quando configurado.

    Raises:
        ValueError: Caso `DATA_SOURCE` não seja "csv" nem "yfinance".
    """
    if DATA_SOURCE == "csv":
        df = pd.read_csv(DATA_PATH)
        if TRAIN_MAX_ROWS and TRAIN_MAX_ROWS > 0:
            df = df.tail(TRAIN_MAX_ROWS).reset_index(drop=True)
        return df


    if DATA_SOURCE == "yfinance":
        import yfinance as yf


        df = yf.download(SYMBOL, start=START_DATE, end=END_DATE, progress=False)
        df = df.reset_index()
        if TRAIN_MAX_ROWS and TRAIN_MAX_ROWS > 0:
            df = df.tail(TRAIN_MAX_ROWS).reset_index(drop=True)
        return df


    raise ValueError(f"DATA_SOURCE inválido: {DATA_SOURCE}. Use 'csv' ou 'yfinance'.")




df = load_data()
print("Loaded DATA_PATH:", DATA_PATH, "shape:", df.shape)
df.head()
```

### Prepare series

```python
def prepare_series(df: pd.DataFrame, date_col: str, target_col: str) -> pd.Series:
    """
    Constrói uma série temporal 1D a partir de um DataFrame tabular.

    Contexto:
        Valida colunas de data e alvo, ordena cronologicamente, remove
        valores inválidos e configura o índice datetime para uso em modelos.

    Args:
        df: DataFrame de entrada contendo colunas de data e alvo.
        date_col: Nome da coluna que representa a data/tempo.
        target_col: Nome da coluna alvo a ser prevista.

    Returns:
        Série temporal (`pd.Series`) com índice datetime e valores float
        prontos para modelagem.

    Raises:
        ValueError: Se `date_col` ou `target_col` não existirem no DataFrame.
    """
    df = df.copy()


    if date_col not in df.columns:
        raise ValueError(f"Coluna de data '{date_col}' não encontrada no DataFrame.")


    if target_col not in df.columns:
        raise ValueError(f"Coluna alvo '{target_col}' não encontrada no DataFrame.")


    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
    df = df.dropna(subset=[date_col, target_col])


    df = df.sort_values(date_col).drop_duplicates(subset=[date_col], keep="last")
    series = df[target_col].astype(float)


    series.index = df[date_col].values
    series.name = target_col
    return series




series = prepare_series(df, DATE_COL, TARGET_COL)
series.head(), series.tail()
```

### Temporal split (train/val/test)

```python
def temporal_split(
    series: pd.Series,
    train_ratio: float,
    val_ratio: float,
    test_ratio: float,
) -> Tuple[pd.Series, pd.Series, pd.Series]:
    """
    Divide a série temporal em subconjuntos de treino, validação e teste.

    Contexto:
        Realiza um split temporal estrito (sem embaralhar) para evitar
        vazamento de informação entre os conjuntos de dados.

    Args:
        series: Série temporal completa em ordem cronológica.
        train_ratio: Proporção de observações alocadas ao treino.
        val_ratio: Proporção de observações alocadas à validação.
        test_ratio: Proporção de observações alocadas ao teste.

    Returns:
        Tupla `(train, val, test)` com três séries temporais disjuntas.

    Raises:
        ValueError: Se a série for muito curta ou se os ratios não somarem 1.0.
    """
    n = len(series)
    if n < 10:
        raise ValueError("Série muito curta para split/treino.")


    if not np.isclose(train_ratio + val_ratio + test_ratio, 1.0):
        raise ValueError("Ratios devem somar 1.0")


    train_end = int(n * train_ratio)
    val_end = int(n * (train_ratio + val_ratio))


    train = series.iloc[:train_end]
    val = series.iloc[train_end:val_end]
    test = series.iloc[val_end:]
    return train, val, test




train_s, val_s, test_s = temporal_split(series, TRAIN_RATIO, VAL_RATIO, TEST_RATIO)
len(train_s), len(val_s), len(test_s)
```

### Scaling

```python
def fit_scaler_on_train(train: pd.Series) -> MinMaxScaler:
    """
    Ajusta um `MinMaxScaler` usando exclusivamente o conjunto de treino.

    Contexto:
        Evita vazamento de informação ao estimar parâmetros de escala
        apenas com dados de treino antes de transformar validação e teste.

    Args:
        train: Série temporal de treino em escala original.

    Returns:
        Instância de `MinMaxScaler` ajustada aos valores de treino.
    """
    scaler = MinMaxScaler()
    scaler.fit(train.values.reshape(-1, 1))
    return scaler




def scale_series(s: pd.Series, scaler: MinMaxScaler) -> np.ndarray:
    """
    Aplica um scaler pré-ajustado a uma série temporal.

    Contexto:
        Normaliza a série para o intervalo definido pelo `MinMaxScaler`,
        retornando um vetor NumPy 1D para uso em janelamento e modelagem.

    Args:
        s: Série temporal em escala original.
        scaler: Instância de `MinMaxScaler` já ajustada.

    Returns:
        Array NumPy 1D contendo os valores escalonados da série.
    """
    return scaler.transform(s.values.reshape(-1, 1)).reshape(-1)




scaler = fit_scaler_on_train(train_s)


train_scaled = scale_series(train_s, scaler)
val_scaled = scale_series(val_s, scaler)
test_scaled = scale_series(test_s, scaler)


train_scaled[:3], val_scaled[:3], test_scaled[:3]
```

### Windowing

```python
def create_temporal_window_1d(values_1d: np.ndarray, window_size: int) -> Tuple[np.ndarray, np.ndarray]:
    """
    Cria janelas temporais 1D para previsão do próximo valor da série.

    Contexto:
        Constrói pares (X, y) onde cada amostra X contém `window_size`
        observações passadas e y é o próximo valor imediato da série.

    Args:
        values_1d: Array 1D com a série já escalonada.
        window_size: Tamanho da janela de histórico em passos de tempo.

    Returns:
        Tupla `(X, y)` onde:
            - X: Array NumPy de shape `(n_amostras, window_size, 1)`;
            - y: Array NumPy de shape `(n_amostras, 1)` com os targets.
    """
    X, y = [], []
    for i in range(window_size, len(values_1d)):
        X.append(values_1d[i - window_size : i])
        y.append(values_1d[i])
    X = np.array(X).reshape(-1, window_size, 1)
    y = np.array(y).reshape(-1, 1)
    return X, y




def make_datasets(
    train_scaled: np.ndarray,
    val_scaled: np.ndarray,
    test_scaled: np.ndarray,
    window_size: int,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray, np.ndarray, np.ndarray, np.ndarray]:
    """
    Monta datasets janelados de treino, validação e teste com contexto correto.

    Contexto:
        Garante que validação e teste enxerguem apenas histórico passado,
        concatenando fatias de treino/validação conforme necessário.

    Args:
        train_scaled: Série de treino escalonada (1D).
        val_scaled: Série de validação escalonada (1D).
        test_scaled: Série de teste escalonada (1D).
        window_size: Tamanho da janela temporal em passos de tempo.

    Returns:
        Tupla `(X_train, y_train, X_val, y_val, X_test, y_test)` contendo
        os arrays NumPy janelados para cada partição.
    """
    X_train, y_train = create_temporal_window_1d(train_scaled, window_size)


    val_concat = np.concatenate([train_scaled[-window_size:], val_scaled])
    X_val, y_val = create_temporal_window_1d(val_concat, window_size)


    test_context = np.concatenate([train_scaled, val_scaled])[-window_size:]
    test_concat = np.concatenate([test_context, test_scaled])
    X_test, y_test = create_temporal_window_1d(test_concat, window_size)


    return X_train, y_train, X_val, y_val, X_test, y_test




X_train, y_train, X_val, y_val, X_test, y_test = make_datasets(
    train_scaled, val_scaled, test_scaled, window_size=HYPERPARAM_GRID["window_size"][0]
)
X_train.shape, X_val.shape, X_test.shape
```

### Model factory

```python
def build_lstm_model(
    window_size: int,
    lstm_units: int,
    dropout_rate: float,
    learning_rate: float = 1e-3,
) -> tf.keras.Model:
    """
    Cria e compila um modelo LSTM sequencial para série temporal 1D.

    Contexto:
        Implementa uma arquitetura simples de duas camadas LSTM com dropout
        e camada densa final para regressão de valor único (próximo passo).

    Args:
        window_size: Número de passos de tempo de entrada por amostra.
        lstm_units: Número de unidades na primeira camada LSTM.
        dropout_rate: Taxa de dropout aplicada após cada camada LSTM.
        learning_rate: Taxa de aprendizado do otimizador Adam.

    Returns:
        Instância compilada de `tf.keras.Model` pronta para treinamento.
    """
    model = Sequential(
        [
            LSTM(lstm_units, return_sequences=True, input_shape=(window_size, 1)),
            Dropout(dropout_rate),
            LSTM(max(lstm_units // 2, 8)),
            Dropout(dropout_rate),
            Dense(1),
        ]
)


    opt = tf.keras.optimizers.Adam(learning_rate=learning_rate)
    model.compile(optimizer=opt, loss="mse")
    return model
```

```python
def inverse_1d(scaler: MinMaxScaler, arr_2d: np.ndarray) -> np.ndarray:
    """
    Reverte a normalização de um array 2D escalonado para a escala original.

    Contexto:
        Converte previsões e targets de volta para os valores reais da
        série temporal após uso de `MinMaxScaler`.

    Args:
        scaler: Instância de `MinMaxScaler` usada no escalonamento.
        arr_2d: Array NumPy 2D de shape `(n, 1)` com valores escalonados.

    Returns:
        Array NumPy 1D contendo os valores na escala original.
    """
    return scaler.inverse_transform(arr_2d).reshape(-1)




def eval_regression_real(y_true_real: np.ndarray, y_pred_real: np.ndarray) -> Dict[str, float]:
    """
    Calcula métricas de regressão (MAE, RMSE, MAPE) na escala real.

    Contexto:
        Avalia o desempenho do modelo em termos absolutos, usando valores
        desnormalizados para facilitar a interpretação dos erros.

    Args:
        y_true_real: Array 1D com valores reais observados.
        y_pred_real: Array 1D com valores previstos pelo modelo.

    Returns:
        Dicionário com chaves `mae`, `rmse` e `mape` em float.
    """
    mae = mean_absolute_error(y_true_real, y_pred_real)
    rmse = np.sqrt(mean_squared_error(y_true_real, y_pred_real))
    mape = mean_absolute_percentage_error(y_true_real, y_pred_real)
    return {"mae": float(mae), "rmse": float(rmse), "mape": float(mape)}




results: List[Dict[str, Any]] = []


for window_size in HYPERPARAM_GRID["window_size"]:
    X_train, y_train, X_val, y_val, _, _ = make_datasets(
        train_scaled, val_scaled, test_scaled, window_size=window_size
)


    if X_train.shape[0] == 0 or X_val.shape[0] == 0:
        print(
            f"[WARN] Sem dados suficientes para window_size={window_size}. ",
            f"X_train.shape={X_train.shape}, X_val.shape={X_val.shape}",
,
```

### Save best model configs (.pkl)

```python
def save_artifact_pkl(
    path: str,
    model: tf.keras.Model,
    scaler: MinMaxScaler,
    window_size: int,
    feature_names: List[str],
    best_params: Dict[str, Any],
    extra_metadata: Optional[Dict[str, Any]] = None,
) -> None:
    """Serializa modelo + scaler + metadados em um arquivo .pkl (joblib)."""
    artifact = {
        "scaler": scaler,
        "window_size": int(window_size),
        "feature_names": feature_names,
        "model_json": model.to_json(),
        "model_weights": model.get_weights(),
        "best_params": best_params,
        "compile": {
            "loss": "mse",
            "optimizer": "adam",
        },
        "metadata": {
            "tf_version": tf.__version__,
            **(extra_metadata or {}),
        },
    }
    joblib.dump(artifact, path)


save_artifact_pkl(
    path=ARTIFACT_PATH,
    model=final_model,
    scaler=scaler,
    window_size=best_params["window_size"],
    feature_names=[TARGET_COL],
    best_params=best_params,
    extra_metadata={"target_col": TARGET_COL},
)

print("Saved:", ARTIFACT_PATH)
```

### Load artifact & Predict (inference)

```python
def load_artifact_pkl(path: str) -> Dict[str, Any]:
    """Carrega um artifact .pkl e reconstrói o modelo Keras."""
    artifact = joblib.load(path)

    model = tf.keras.models.model_from_json(artifact["model_json"])
    model.set_weights(artifact["model_weights"])

    model.compile(optimizer="adam", loss="mse")

    artifact["model"] = model
    return artifact


def predict_next_from_recent(
    artifact: Dict[str, Any],
    recent_values: Union[pd.Series, np.ndarray],
) -> float:
    """Prevê o próximo valor real usando a última janela de valores reais."""
    scaler: MinMaxScaler = artifact["scaler"]
    window_size: int = artifact["window_size"]
    model: tf.keras.Model = artifact["model"]

    if isinstance(recent_values, pd.Series):
        recent = recent_values.values.astype(float)
    else:
        recent = np.asarray(recent_values, dtype=float)

    if len(recent) < window_size:
        raise ValueError(f"Precisa de pelo menos {window_size} valores recentes.")

    last_window = recent[-window_size:].reshape(-1, 1)
    last_window_scaled = scaler.transform(last_window).reshape(1, window_size, 1)

    pred_scaled = model.predict(last_window_scaled, verbose=0).reshape(-1, 1)
    pred_real = scaler.inverse_transform(pred_scaled).reshape(-1)[0]
    return float(pred_real)


artifact = load_artifact_pkl(ARTIFACT_PATH)

example_recent = test_s.iloc[-artifact["window_size"] :]
next_pred = predict_next_from_recent(artifact, example_recent)

next_pred

with open(OUTPUT_PATH, "w") as f:
    json.dump({"prediction_r$": next_pred}, f)
print(f"Salvo em {OUTPUT_PATH}")
```
